```python
import pandas as pd
```

```python
# 读取数据
data = pd.read_csv('../../../data/advertising.csv')
print(data.head())
print(data.shape)
```

```python
# 数据预处理，去掉第一列
data_filtered = data.iloc[:,1:]
print(data_filtered.shape)
```

```python
# 定义特征 X 和标签列
X = data_filtered.iloc[:,:-1]
y = data_filtered.iloc[:,-1]
```

```python
# 计算皮尔逊相关系数
r = X.corrwith(y, method='pearson')
print(r)
```

```python
# 皮尔逊相关系数矩阵
corr_matrix = data_filtered.corr(method='pearson')
print(corr_matrix)
```
