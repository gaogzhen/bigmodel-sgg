```python
import numpy as np
```

```python
# 1. 均匀分布
a = np.random.rand(5)
print(a)
```

```python
a = np.random.rand(2, 3)
print(a)
```

```python
b = np.random.randint(0, 100, 5)
print(b)
```

```python
b = np.random.randint(0, 100, size=(3, 4))
print(b)
```

```python
# 2. 正态分布
c = np.random.randn(10)    # 标准正态分布
print(c)
```

```python
c = np.random.randn(2, 5)    # 标准正态分布
print(c)
```

```python
print(c * 10 + 5)
```

```python
d = np.random.normal(5, 10, size=(2,5))
print(d)
```
