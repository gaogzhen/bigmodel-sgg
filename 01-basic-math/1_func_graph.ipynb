```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 画出y=sinx的函数图像
x = np.arange(0, 6, 0.1)
y = np.sin(x)
print(x)
print(y)
```

```python
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sinx")
plt.show()
```

```python
# 画出导函数 y' = cosx 的图像
y1 = np.cos(x)
plt.plot(x, y1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sin'x")
plt.show()
```

```python
# 将函数和导函数图像放在同一个图中
plt.plot(x, y, label="y=sinx")
plt.plot(x, y1, label="y'=cosx", linestyle="--")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sinx")
plt.legend()
plt.show()
```
