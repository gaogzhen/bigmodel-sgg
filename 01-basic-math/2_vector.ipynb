```python
import numpy as np
```

```python
x = np.array([2,5,8])
print(x)
print(x.shape)
```

```python
# 1. 转置
print(x.T)
print(x.T.shape)
```

```python
# 2. 向量相加
y = np.array([1,3,7])
print(x + y)
```

```python
# 3. 向量与标量相乘
print(x * 3)
```

```python
# 4. 向量与向量相乘（对位相乘）
print(x * y)
```

```python
# 5. 向量点乘
print(x.dot(y))
print(np.dot(x, y))
print(x @ y)
```

```python
# 6. 计算范数
l0_norm = np.linalg.norm(x, ord=0)
print(l0_norm)
```

```python
l1_norm = np.linalg.norm(x, ord=1)
print(l1_norm)
l2_norm = np.linalg.norm(x, ord=2)
print(l2_norm)
lp_norm = np.linalg.norm(x, ord=np.inf)
print(lp_norm)
```
