```python
import numpy as np
```

基本矩阵运算

```python
A = np.array([[1,2,3],[4,5,6]])
B = np.array([[1,3,7],[5,0,2]])
print(A)
print(A.shape)
print(B)
print(B.shape)
```

```python
# 1. 取矩阵中的某个元素
print(A[0][1])
```

```python
print(A[1])
```

```python
# 2. 转置
print(A.T)
print(A.T.shape)
```

```python
# 3. 矩阵相加，需要形状相同
print(A + B)
```

```python
C = np.array([[1,3,7],[5,0,2],[5,6,7]])
print(C.shape)
```

```python
# 形状不同不能相加
# print(A + C)
```

```python
# 矩阵和向量相加
x = np.array([2,5,8])
print(A + x)
```

```python
# 4. 矩阵与标量相乘
print(A * 3)
```

```python
# 5. 矩阵与向量相乘（向量广播后对位相乘）
print(A * x)
```

```python
# 6. 矩阵与矩阵相乘（对位相乘）
print(A * B)
```

```python
# 7. 矩阵乘法
print(A.dot(B.T))
```

```python
print(A @ C)
```

```python
# 8. 矩阵求逆
D = np.array([[1,2],[3,5]])
D_inv = np.linalg.inv(D)
print(D_inv)
```

```python
print(D @ D_inv)
print(D_inv @ D)
```

```python
D_inv = np.around(D_inv).astype(int)
print(D_inv)
```

```python
print(D @ D_inv)
print(D_inv @ D)
```

```python
A_inv = np.linalg.pinv(A)
print(A)
print(A_inv)
```

```python
print(A @ A_inv)
print(A_inv @ A)
```

其它矩阵运算

```python
# 9. 向量化
print(A.flatten())
```

```python
# 10. 哈达玛积（对位相乘）
print(A * B)
print(np.multiply(A, B))
```

```python
# 11. 矩阵内积
print(np.sum(A * B))
print(np.dot(A.flatten(), B.flatten()))
```

```python
# 12. 克罗内克积
print(A)
print(B)
print(np.kron(A, B))
```

张量的定义和形状

```python
tensor = np.array([[[1,2],[3,5],[4,8]],[[1,2],[3,5],[4,8]],[[1,2],[3,5],[4,8]]])
print(tensor)
print(tensor.shape)
```

梯度计算

```python
f = np.array([1,2,4,7,11,16])
print(np.gradient(f))
```
